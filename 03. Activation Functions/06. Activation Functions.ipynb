### Activation Functions

Activation functions are mathematical functions applied to the output of neurons.
They transform the neuron’s output before passing it to the next layer.

Activation functions are applied to neuron outputs to introduce non-linearity into neural networks.
They determine whether a neuron “fires” and how strongly it contributes to the next layer.

### Purpose

Activation functions introduce non-linearity into the network.

Without them, the entire neural network behaves like a linear model, no matter how many layers it has.

Non-linear activation allows the network to learn complex patterns and relationships in data.

### Types of ***Activation Functions***

* ***Hidden layer activations*** — used in internal layers (e.g., ReLU, tanh, sigmoid)

Usually the same across all hidden layers

Help capture non-linear relationships

* ***Output layer activations*** — used depending on the task type

For classification, you might use Softmax

For regression, often no activation (linear output)

## Why Use Activation Functions?

- In most cases, for a
neural network to fit a nonlinear function, we need it to contain two or more hidden layers, and
we need those hidden layers to use a nonlinear activation function.

- First off, what’s a nonlinear function? A nonlinear function cannot be represented well by a
straight line, such as a sine function: Non-Linear

- While there are certainly problems in life that are linear in nature, for example, trying to figure
out the cost of some number of shirts, and we know the cost of an individual shirt, and that there
are no bulk discounts, then the equation to calculate the price of any number of those products is a
linear equation.

- The number of
factors that come into play, such as size, location, time of year attempting to sell, number of
rooms, yard, neighborhood, and so on, makes the pricing of a home a nonlinear equation
